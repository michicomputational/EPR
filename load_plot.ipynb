## load & plot

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
import os
import numpy as np

folder = "/home/michael_lmu/EPR/saved_results/network epr/fake feed-forward epr/"
data = {}

for filename in os.listdir(folder):
    if filename.endswith(".npy"):
        filepath = os.path.join(folder, filename)
        # allow_pickle=True is needed if they contain Python objects like dicts
        data[os.path.splitext(filename)[0]] = np.load(filepath, allow_pickle=True).item()

# Now you can access each dict by its filename (without extension)
print(data.keys())       # e.g. dict_keys(['file1', 'file2'])
#print(data["fff_33Hz_epr5runs_dt1_expflt5_norm__epochmax20000_dimh5_numblocks2"])     # dictionary loaded from file1.npy
#print(data["fff_33Hz_epr5runs_dt1_expflt15_norm__epochmax20000_dimh5_numblocks2"])     # dictionary loaded from file1.npy
#print(data["fff_33Hz_epr5runs_dt1_expflt25_norm__epochmax20000_dimh5_numblocks2"])     # dictionary loaded from file1.npy
#print(data["fff_33Hz_pois-ctrl_epr5runs_dt1_expflt5_norm__epochmax20000_dimh5_numblocks2"])     # dictionary loaded from file1.npy
#print(data["fff_33Hz_pois-ctrl_epr5runs_dt1_expflt15_norm__epochmax20000_dimh5_numblocks2"])     # dictionary loaded from file1.npy
```

```python
#exponential filters
flt05 = data["fff_33Hz_epr5runs_dt1_expflt5_norm__epochmax20000_dimh5_numblocks2"]    # dictionary loaded from file1.npy
flt10 = data['fff_33Hz_epr5runs_dt1_expflt10_norm__epochmax20000_dimh5_numblocks2']
flt15 = data["fff_33Hz_epr5runs_dt1_expflt15_norm__epochmax20000_dimh5_numblocks2"]     # dictionary loaded from file1.npy
flt15 = data["fff_33Hz_epr5runs_dt1_expflt15_norm__epochmax20000_dimh5_numblocks2"]     # dictionary loaded from file1.npy
flt20 = data["fff_av33.34Hz_epr5runs_dt1_expflt20_norm__epochmax20000_dimh5_numblocks2"]     # dictionary loaded from file1.npy
flt25 = data["fff_33Hz_epr5runs_dt1_expflt25_norm__epochmax20000_dimh5_numblocks2"]     # dictionary loaded from file1.npy
flt30 = data["fff_av33.34Hz_epr5runs_dt1_expflt30_norm__epochmax20000_dimh5_numblocks2"]     # dictionary loaded from file1.npy
flt35 = data["fff_33Hz_epr5runs_dt1_expflt35_norm__epochmax20000_dimh5_numblocks2"]     # dictionary loaded from file1.npy

flt05_ctrl = data["fff_33Hz_pois-ctrl_epr5runs_dt1_expflt5_norm__epochmax20000_dimh5_numblocks2"]     # dictionary loaded from file1.npy
flt15_ctrl = data["fff_33Hz_pois-ctrl_epr5runs_dt1_expflt15_norm__epochmax20000_dimh5_numblocks2"]     # dictionary loaded from file1.npy
explist = [flt05, flt10, flt15, flt20, flt25, flt30, flt35]
explist_flt = [5, 10, 15, 20, 25, 30, 35]
explist_ctrl = [flt05_ctrl, flt15_ctrl]

#gaussian filters
flt05_gauss = data['fff_33Hz_epr5runs_dt1_gaussflt5_norm__epochmax20000_dimh5_numblocks2']
flt10_gauss = data['fff_33Hz_epr5runs_dt1_gaussflt10_norm__epochmax20000_dimh5_numblocks2']
flt15_gauss = data['fff_33Hz_epr5runs_dt1_gaussflt15_norm__epochmax20000_dimh5_numblocks2']
flt05_gauss_ctrl = data['fff_33Hz_pois-ctrl_epr5runs_dt1_gaussflt5_norm__epochmax20000_dimh5_numblocks2']
flt15_gauss_ctrl = data['fff_33Hz_pois-ctrl_epr5runs_dt1_expflt5_norm__epochmax20000_dimh5_numblocks2']

gausslist = [flt05_gauss, flt10_gauss, flt15_gauss]
gausslist_flt = [5, 10, 15]
gausslist_ctrl = [flt05_gauss_ctrl, flt15_gauss_ctrl]


#alpha filter
flt03_alph = data["fff_av33.34Hz_epr5runs_dt1_alphaflt3_norm__epochmax20000_dimh5_numblocks2"]     # dictionary loaded from file1.npy
flt05_alph = data["fff_av33.34Hz_epr5runs_dt1_alphaflt5_norm__epochmax20000_dimh5_numblocks2"]     # dictionary loaded from file1.npy
flt10_alph = data["fff_av33.34Hz_epr5runs_dt1_alphaflt10_norm__epochmax20000_dimh5_numblocks2"]     # dictionary loaded from file1.npy
flt15_alph = data["fff_av33.34Hz_epr5runs_dt1_alphaflt15_norm__epochmax20000_dimh5_numblocks2"]     # dictionary loaded from file1.npy
flt20_alph = data["fff_av33.34Hz_epr5runs_dt1_alphaflt20_norm__epochmax20000_dimh5_numblocks2"]     # dictionary loaded from file1.npy

alphalist = [flt03_alph, flt05_alph, flt10_alph, flt15_alph, flt20_alph]
alphalist_flt = [3,5,10,15,20]



```

```python
fig,ax = plt.subplots(1,3,figsize = (9,5),sharey=True)

#exp filter
for i in range(len(explist)):
    epr = explist[i]['EPR_test'][:,-1]
    flt = explist_flt[i]
    ax[0].scatter(np.repeat(flt, 5), epr, color='lightgrey')
    ax[0].errorbar(flt, np.mean(epr), yerr=np.std(epr),color='black',fmt='o')
ax[0].set_title('exp-filter')
ax[0].set_xlabel('time constant [ms]')
ax[0].set_ylabel('EPR')


#gauss
for i in range(len(gausslist)):
    epr = gausslist[i]['EPR_test'][:,-1]
    flt = gausslist_flt[i]
    ax[1].scatter(np.repeat(flt, 5), epr, color='lightgrey')
    ax[1].errorbar(flt, np.mean(epr), yerr=np.std(epr),color='black',fmt='o')
ax[1].set_title('gauss-filter')
ax[1].set_xlabel('std [ms]')

#alpha filter
for i in range(len(alphalist)):
    epr = alphalist[i]['EPR_test'][:,-1]
    flt = alphalist_flt[i]
    ax[2].scatter(np.repeat(flt, 5), epr, color='lightgrey')
    ax[2].errorbar(flt, np.mean(epr), yerr=np.std(epr),color='black',fmt='o')
ax[2].set_title('alpha-filter')
ax[2].set_xlabel('time constant [ms]')


plt.suptitle('EPR of 3-neuron feed-forward activity for different filter-types and -sizes')
```
